A directly sourced AnsDatabase file has all of the **component groups** in its **source group** contain values and does not refer to any external files.

In this example we build a directly sourced file from the default HITRAN AnsDatabase file.

```python
from pathlib import Path

import h5py

import archnemesis as ans
import archnemesis.helpers.h5py_helper as h5py_helper
from archnemesis.database.filetypes.ans_partition_fn_data_file import AnsPartitionFunctionDataFile
from archnemesis.database.filetypes.ans_line_data_file import AnsLineDataFile

# Set path to default AnsDatabase file
ans_database_pf_default = AnsPartitionFunctionDataFile(Path(ans.Data.path_data.archnemesis_path()) / "archnemesis/Data/partition_functions/tips2025.h5")
ans_database_ld_default = AnsLineDataFile(Path(ans.Data.path_data.archnemesis_path()) / "docs/examples/linedata/test_data/HITRAN2024_DIGITAL_CSIC/hitran24.h5")

# Set path to new direct source file
ans_database_direct_source_path = Path(ans.Data.path_data.archnemesis_path()) / "docs/examples/linedata/test_data/direct_source.h5"
if ans_database_direct_source_path.exists():
	ans_database_direct_source_path.unlink()
ans_database_pf_direct_source = AnsPartitionFunctionDataFile(ans_database_direct_source_path)
ans_database_ld_direct_source = AnsLineDataFile(ans_database_direct_source_path)

# Choose some molecules to directly source

mol_names = (
	'H2O',
	'CH4',
	'CO2',
)


for mol_name in mol_names:
	print(f'{mol_name=}')
	for src_mol, src_mol_grp in ans_database_pf_default.iter_mol_grps():
		if src_mol == mol_name:
			ans_database_pf_direct_source.add_source_data_grp(ans_database_pf_default.path.stem, src_mol_grp, description=f'Direct source group from {ans_database_pf_default.path!s}')
	
	for src_mol, src_mol_grp in ans_database_ld_default.iter_mol_grps():
		if src_mol == mol_name:
			ans_database_ld_direct_source.add_source_data_grp(ans_database_ld_default.path.stem, src_mol_grp, description=f'Direct source group from {ans_database_ld_default.path!s}')
```
